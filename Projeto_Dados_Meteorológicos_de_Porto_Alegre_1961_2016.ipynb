```python
from os import listxattr
#Parte 1.A
#Lista vazia para guardar dos dados.
dados_meteorologicos_documento=[]

#Lê o arquivo.
arquivo=open("Anexo_Arquivo_Dados_Projeto_Logica_e_programacao_de_computadores.csv", "r")

#Lê a primeira linha e reserva o cabeçalho separadamente, evita de atrapalhar os cálculos e dar erro.
cabecalho=arquivo.readline()

#Faz uma repetição para ler o restante das linhas.
for linha in arquivo:

  #Limpa o último caractere invisível no final e divide os dados pelas vírgulas.
  linha_invisivel_limpa=linha [:-1]
  dados_divididos=linha_invisivel_limpa.split(",")

  #Adiciona os dados divididos desse dia na lista.
  dados_meteorologicos_documento.append(dados_divididos)

#Fecha o arquivo.
arquivo.close()

#Imprime as 5 primeiras linhas para testar.
print(dados_meteorologicos_documento[:5])

#Parte 2.A
#Pegar e validar o mês inicial.
while True:
  mes_inicial= int(input("Digite o mês inicial (1 a 12):"))
  if 1 <=mes_inicial <=12:
    break
  else:
    print("Erro: Mês inválido. Por favor, digite um número entre 1 e 12.")

#Pegar e validar o ano inicial.
while True:
  ano_inicial= int(input("Digite o ano inicial (1961 a 2016):"))
  if 1961<=ano_inicial <= 2016:
    break
  else:
    print("Erro: Ano inválido. Os anos disponíveis são entre 1961 e 2016.")

#Pegar e validar o mês final.
while True:
  mes_final= int(input("Digite o mês final (1 a 12):"))
  if 1 <=mes_final <=12:
    break
  else:
    print("Erro: Mês inválido. Por favor, digite um número entre 1 e 12.")

#Pegar e validar o ano final.
while True:
  ano_final= int(input("Digite o ano final (1961 a 2016):"))
  if 1961<=ano_final <=2016:
    break
  else:
    print("Erro: Ano inválido. Os anos disponíveis são entre 1961 e 2016.")

#Pegar e validar a opção de visualização.
while True:
  print("\n--- Opções de Vizualização ---")
  print("1- Todos os dados")
  print("2- Apenas precipitação")
  print("3- Apenas temperatura")
  print("4- Apenas umidade e vento")

  opcao= int(input("Digite o número da opção desejada (1 a 4):"))

  if 1<=opcao <=4:
    break
  else:
    print("Erro: Opção inváçida. Escolha um número entre 1 e 4.")

#Parte 3.A
#Cabeçalho.
print("\n" + "="*60)
print("Resultados da Pequisa Meteorológica.")
print("="*60)

periodo_inicial=(ano_inicial*100) + mes_inicial
periodo_final= (ano_final*100) + mes_final

#Passar por toda a lista de dados do arquivo.
for dia in dados_meteorologicos_documento:
  data_completa= dia[0]

  #Dividir as datas com barras para pegar o mês e ano corretamente.
  partes_data= data_completa.split("/")
  mes_do_dia= int(partes_data[1])
  ano_do_dia= int(partes_data[2])

  #Criar o 'código' para o dia atual.
  periodo_atual=(ano_do_dia*100) + mes_do_dia

  #Verificar se o dia atual está dentro do período desejado pelo usuário.
  if periodo_inicial<= periodo_atual<= periodo_final:

    #Se foi, ver o que deve imprimir de acordo com a opção 1 a 4.
    if opcao ==1: #Todos os dados.
      print(f"Data:{dia[0]} | Precipitação:{dia [1]} | Temp.Máxima:{dia[2]} | Tem.Mínima:{dia[3]} | Umidade:{dia[6]} | Vento:{dia[7]}")
    elif opcao ==2: #precipitação.
      print(f"Data:{dia[0]} | Precipitação;{dia[1]}")
    elif opcao ==3: #Temperatura.
      print(f"Data:{dia[0]} | Temp. Máxima:{dia[3]} | Temp.Mínima:{dia[4]}")
    elif opcao ==4: #Umidade e vento.
      print(f"Data:{dia[0]} | Umidade:{dia[6]} | Vento{dia[7]}")

print("="*60)

#Colocar meses por extenso. (0 vazio)
meses_por_extenso= ["","janeiro","fevereiro","março","abril","maio","junho","julho","agosto","setembro","outubro","novembro","dezembro"]

def calcular_mes_mais_chuvoso(dados):

    chuva_por_mes_ano={}

    for dia in dados:
        data_completa=dia[0]

        #Caso tenha dias sem nenhum dado será usado o  try/except para não dar pane.
        try:
            precipitacao= float(dia[1])
        except ValueError:
          continue

        #Dividir a data para pegar o mês e o ano.
        partes_data=data_completa.split("/")
        mes= partes_data [1]
        ano= partes_data[2]

        #Criação chave mês/ano.
        chave_mes_ano= f"{mes}/{ano}"

        #Se já tem o mês/ano, soma a chuva do dia no total dele.
        if chave_mes_ano in chuva_por_mes_ano:
            chuva_por_mes_ano[chave_mes_ano] = chuva_por_mes_ano[chave_mes_ano] + precipitacao

        #Se não tem o mês/ano,cria a chave no dicionário já com a chuva deste primeiro dia.
        else:
            chuva_por_mes_ano[chave_mes_ano] = precipitacao

    #Descobrir o mês/ano mais chuvoso.
    maior_volume_chuva=0.0
    mes_ano_mais_chuvoso=""

    #Analisar o dicionário para achar o maior valor.
    for chave, volume in chuva_por_mes_ano.items():
       if volume > maior_volume_chuva:
          maior_volume_chuva=volume
          mes_ano_mais_chuvoso=chave

    #Função com a data com mais chuva e volume total da mesma.
    return mes_ano_mais_chuvoso, maior_volume_chuva

print("\n" + "="*60)
print("Resultado: Mês mais chuvoso (1961-2016)")
print("="*60)

#Função verificando a lista de dados.
data_mais_chuvosa, chuva_maxima= calcular_mes_mais_chuvoso (dados_meteorologicos_documento)

#Pega o resultado e divide para pegar o número do mês e ano separados.
partes_data_mais_chuvosa= data_mais_chuvosa.split("/")
numero_mes_mais_chuvoso= int(partes_data_mais_chuvosa[0])
ano_mais_chuvoso=partes_data_mais_chuvosa[1]

#Usa a lista para pegar o nome do mês por extenso.
nome_do_mes=meses_por_extenso[numero_mes_mais_chuvoso]

#Imprime o resultado final na tela.
print(f"O mês mais chuvoso foi {nome_do_mes} de {ano_mais_chuvoso} com {chuva_maxima:.2f} mm de precipitação.")
print("="*60)

#Parte 1.C
#Média da temperatura mínima de um determinado mês entre 2006 e 2016.

#Lista meses por extenso (0 vazio).
meses_por_extenso = ["", "janeiro", "fevereiro", "março", "abril", "maio", "junho", "julho", "agosto", "setembro", "outubro", "novembro", "dezembro"]

#Validação da entrada do usuário.
while True:
  mes_informado=int(input("\nDigite o número do mês para analisar a temperatura mínima (1 a 12):"))
  if 1 <= mes_informado <= 12:
    break
  else:
    print("Erro:Mês inválido. Por favor, digite um número entre 1 e 12.")

#Função.
def calcular_media_minima(dados,mes_alvo):

  #Dicionários temporários
  somas_temperatura_minima={}
  contagem_dias={}

  #Passa pela lista de dados.
  for dia in dados:
    data_completa=dia[0]

    #Dividir a data nas barras.
    partes_data=data_completa.split("/")
    mes_do_dia=int(partes_data[1])
    ano_do_dia=int(partes_data[2])

    #Apenas o mês que o usuário quer entre os anos de 2006 a 2016.
    if mes_do_dia == mes_alvo and 2006 <=ano_do_dia <= 2016:

      #Temperatura mínima, índice 3.
      try:
        temperatura_minima=float(dia[3])
      except ValueError:
          continue #célula vazia, passa para o próximo dia

      #2.C
      #Chave do dicionário, nome do mês + ano.
      nome_mes=meses_por_extenso[mes_alvo]
      chave_dicionario=f"{nome_mes}{ano_do_dia}"

      #Se já tem a chave,soma a temperatura e aumenta a contagem de dias.
      if chave_dicionario in somas_temperatura_minima:
                somas_temperatura_minima[chave_dicionario] += temperatura_minima
                contagem_dias[chave_dicionario] += 1
      # Se a chave não existe ainda, cria com o primeiro valor e 1 dia
      else:
          somas_temperatura_minima[chave_dicionario] = temperatura_minima
          contagem_dias[chave_dicionario] = 1

  #Dicionário final e calcular as médias (soma dividida pela quantidade de dias).
  dicionario_medias={}
  for chave in somas_temperatura_minima:
    media=somas_temperatura_minima[chave] / contagem_dias[chave]
    dicionario_medias[chave] =media

  #Funçao que retorna o dicionário feito.
  return dicionario_medias

print("\n"+"="*60)
print(f"Médias das temperaturas mínimas - {meses_por_extenso[mes_informado].upper()}(2006 a 2016)")
print("="*60)

#Função passa pela lista de dados lidos na 1.A
dicionario_final=calcular_media_minima(dados_meteorologicos_documento,mes_informado)

#Imprime os resultados armazenados no dicionário.
for chave, media in dicionario_final.items():
  print(f"{chave}:{media:.2f} °C")

print("="*60)

#Parte 1.D
#Biblioteca de gráficos.
import matplotlib.pyplot as plt

#Pegando os dados do dicionário que foi criado no item C (.keys() e .values())
periodos=list(dicionario_final.keys())
temperaturas=list(dicionario_final.values())

#Tamanho do gráfico.
plt.figure(figsize=(10,6))

#Gráfico de barras (com label).
plt.bar(periodos,temperaturas,color='royalblue',label='Média Tempertura Mínima (°C)')

#Eixos,título e legenda
nomes_do_mes_grafico=meses_por_extenso[mes_informado].capitalize()
plt.title(f"Médias de Temperatura Mínima - {nomes_do_mes_grafico} (2006 a 2016)")
plt.xlabel("Período (Mês e Ano)")
plt.ylabel("Temperatura Média (°C)")

#Comando pra exibir a legenda.
plt.legend()

#Rotacionando os nomes no eixo X para melhorar a legibilidade.
plt.xticks(rotation=45)

#Comando para exibir o gráfico na tela.
plt.show()

#Parte 1.E
#Média geral da temperatura mína do mês (2006 a 2016)

#Variáveis acumuladoras.
soma_geral=0.0
quantidade_anos=0

#Usar .items() para percorrer o dicionário do item C.
for chave, media_anual in dicionario_final.items():
  soma_geral+=media_anual
  quantidade_anos+=1

#Calcula a média geral
if quantidade_anos > 0:
   media_geral=soma_geral/quantidade_anos
else:
  media_geral=0

#Fazer o nome do mês.
nome_do_mes_escolhido=meses_por_extenso[mes_informado].capitalize()

#Exibindo o resultado na tela.
print("\n"+"="*60)
print(f"Média Geral da Temperatura Mínima - {nome_do_mes_escolhido} (2006 a 2016)")
print("="*60)
print(f"A média geral das temperaturas mínimas para o mês de {nome_do_mes_escolhido} ao longo dos 11 anos foi de {media_geral:.2f} °C.")
print("="*60)





















```
